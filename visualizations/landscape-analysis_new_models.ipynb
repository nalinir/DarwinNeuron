### Visualization

```python
import sqlite3, numpy as np, pandas as pd
import matplotlib.pyplot as plt

def plot_feature():
    # Connect to the SQLite database
    con = sqlite3.connect("data/landscape-analysis.db")
    query = "SELECT * FROM randman_loss_surfaces WHERE ic_h_max IS NOT NULL AND ic_h_max != 'pending'"
    df = pd.read_sql_query(query, con)
    con.close()
    
    # Choose the feature to visualize
    feature = 'ic_eps_s'

    # Get unique problems
    problems = sorted(df['problem_id'].unique())

    # Prepare plot
    fig, axes = plt.subplots(len(problems), 1, figsize=(10, len(problems) * 2), sharex=False)
    
    # Handle case when there's only one problem
    if len(problems) == 1:
        axes = [axes]
    
    # Plot histograms per problem
    for i, problem in enumerate(problems):
        problem_data = df[df['problem_id'] == problem]
        
        # Calculate bins specific to this problem's data
        min_val = df[feature].min()
        max_val = df[feature].max()
        bins = np.linspace(min_val, max_val, 40)
        
        ax = axes[i]
        ax.hist(problem_data[feature], bins=bins, orientation='vertical')
        ax.set_yticks([])  # Hide y-axis ticks for compactness
        ax.set_ylabel(f"Problem {problem}", rotation=0, labelpad=30, va='center')
        
        # Add statistics to help compare
        mean_val = problem_data[feature].mean()
        ax.axvline(mean_val, color='red', linestyle='dashed', linewidth=1, label=f'Mean: {mean_val:.3f}')
        
        ax.spines['top'].set_visible(False)
        ax.spines['right'].set_visible(False)
        ax.spines['left'].set_visible(False)
        ax.legend()

    # Aesthetics
    axes[-1].set_xlabel(feature)
    plt.suptitle(f"Distribution of {feature} by Problem", fontsize=14)
    plt.tight_layout(rect=[0, 0, 1, 0.96])
    plt.show()
plot_feature()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import sqlite3
import numpy as np

def plot_multiple_features(features, where_clause_2=None):
    """
    Connects to a SQLite database, queries data, and plots feature histograms
    for different problems in a grid format.

    The plots are organized with each row representing a unique problem, and
    each column representing a feature. The y-axis labels are dynamically
    set to indicate the Randman ID, model type, and whether it's 'Recurrent'
    or 'Non Recurrent'.
    """
    # Connect to the SQLite database
    try:
        con = sqlite3.connect("data/landscape-analysis.db")
        # SQL query to join two tables and filter for non-null/non-pending data
        # Build a WHERE clause that requires all features to be NOT NULL
        where_clause = " AND ".join([f"{feature} IS NOT NULL" for feature in features])
        query = f"""SELECT a.*, b.randman_id, b.model_type, b.recurrent, b.percent_snn FROM randman_loss_surfaces a
                JOIN randman_problems b on a.problem_id = b.id 
                WHERE {where_clause} {where_clause_2} order by b.recurrent, b.randman_id, b.model_type"""
        df = pd.read_sql_query(query, con)
    except sqlite3.Error as e:
        print(f"Database error: {e}")
        return
    finally:
        if con:
            con.close()

    # Define features to visualize
    # features = ['ic_h_max', 'ic_eps_max', 'ic_eps_s', 'ic_m0', 'ic_eps_ratio']

    # Get unique combinations of randman_id, model_type, and recurrent status
    problems = df[['randman_id', 'model_type', 'recurrent', 'percent_snn']].drop_duplicates().sort_values(by=['randman_id', 'model_type', 'recurrent', 'percent_snn']).values.tolist()
    # Drop the percent_snn != NaN where model_type = NSN
    problems = [problem for problem in problems if not (problem[1].startswith("NSN") and not np.isnan(problem[3]))]
    # Define a remapping dictionary for model type labels
    remapping = {
        "SNN": "SNN",
        "ANN_with_LIF_output": "ANN",
        "NSN_with_LIF_output": "Nonspiking",
        "Hybrid_RNN_SNN_rec": "SNN-ANN (Network-to-Network)",
        "Hybrid_NSN_SNN_rec": "SNN-Nonspiking (Network-to-Network)",
        "Hybrid_RNN_SNN_V1_same_layer": "SNN-ANN (Same Layer)",
        "Hybrid_NSN_SNN_V1_same_layer": "SNN-Nonspiking (Same Layer)",
        "old": "SNN (Evolutionary Team Implementation)",
    }
    
    print(problems)
    # Create subplot grid with shared x-axes for columns
    fig, axes = plt.subplots(len(problems), len(features), 
                             figsize=(len(features) * 3, len(problems) * 1.5), 
                             sharex='col')
    
    # Handle the case where there is only one problem to ensure `axes` is a 2D array
    if len(problems) == 1:
        axes = [axes]
    
    # Plot histograms for each problem and feature combination
    for i, (randman_id, model_type, recurrent, percent_snn) in enumerate(problems):
        
        # Correctly filter the DataFrame for the current unique combination of randman_id, model_type, and recurrent
        if pd.isnull(percent_snn):
            problem_data = df[(df['randman_id'] == randman_id) & 
                      (df['model_type'] == model_type) & 
                      (df['recurrent'] == recurrent) &
                      (df['percent_snn'].isnull())]
        else:
            problem_data = df[(df['randman_id'] == randman_id) & 
                      (df['model_type'] == model_type) & 
                      (df['recurrent'] == recurrent) &
                      (df['percent_snn'] == percent_snn)]
        
        # Find the remapped model type name based on a prefix match
        remapped_model_type = model_type
        for prefix, new_label in remapping.items():
            if model_type.startswith(prefix):
                remapped_model_type = new_label
                break
        
        # Create the conditional label for the y-axis
        recurrent_label = "Recurrent" if recurrent == 1 else "Non Recurrent"
        # print(model_type)
        # print(type(percent_snn))
        # Handle percent_snn label: if it's NaN, set to 0.5 for Hybrid models, else "N/A"
        if np.isnan(percent_snn):
            percent_snn_label = 0.5 if "Hybrid" in model_type else "N/A"
        else:
            percent_snn_label = f"{percent_snn:.2f}"
        # Updated row_label to be three lines for better spacing
        row_label = (f"Randman ID: {randman_id}\n"
                     f"{remapped_model_type}\n"
                     f"Percent SNN {percent_snn_label}\n"
                     f"({recurrent_label}, n={len(problem_data)})")
        # print(row_label)
        # print(len(problem_data))
        for j, feature in enumerate(features):
            ax = axes[i][j]
            
            # Prepare data, dropping NaNs
            problem_feature_data = problem_data[feature].dropna()
            
            # Check if all values for this feature in this problem are the same
            if len(problem_feature_data.unique()) <= 1 and not problem_feature_data.empty:
                unique_val = problem_feature_data.iloc[0]
                # Plot a vertical line and add text for single-value data
                ax.axvline(x=unique_val, color='r', linestyle='--')
                ax.text(unique_val, ax.get_ylim()[1] * 0.8,
                        f'{unique_val:.2f}',
                        ha='center', va='top', rotation=90, fontsize=8, color='r')
            else:
                # Use np.linspace to create consistent bins across all plots for a given feature
                bins = np.linspace(df[feature].min(), df[feature].max(), 40)
                ax.hist(problem_feature_data, bins=bins, orientation='vertical')
            
            # Hide y-axis ticks and spines for a cleaner look
            ax.set_yticks([])
            ax.spines['top'].set_visible(False)
            ax.spines['right'].set_visible(False)
            ax.spines['left'].set_visible(False)
            
            # Set labels only for the first column and last row
            if j == 0:
                ax.set_ylabel(row_label, rotation=0, labelpad=90, va='center')
            
            if i == len(problems) - 1:
                ax.set_xlabel(feature, fontsize=9)
    
    # Add a main title and adjust the layout to prevent overlap
    plt.suptitle("Feature Distributions by Randman Dataset and Model Type", fontsize=14)
    # Adjust subplot params to make space for the longer y-axis labels
    plt.subplots_adjust(left=0.4)
    plt.tight_layout(rect=[0, 0, 1, 0.96])
    plt.show()

# Run the function to generate the plot
plot_multiple_features(
    ['ic_h_max', 'ic_eps_s', 'ic_eps_max', 'ic_m0', 'ic_eps_ratio'], "and b.model_path is not NULL"
)
# This validates what we see with the smoothness for nonspiking
```

```python
from mpl_toolkits.mplot3d import Axes3D
from sklearn.decomposition import PCA
import sqlite3
import umap
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

def plot_randman_pca(features=['ic_h_max', 'ic_eps_s', 'ic_eps_max', 'ic_eps_ratio', 'ic_m0'], other_clause = ""):
    remapping = {
        "SNN": "SNN",
        "ANN_with_LIF_output": "ANN",
        "NSN_with_LIF_output": "Nonspiking",
        "Hybrid_RNN_SNN_rec": "SNN-ANN (Network-to-Network)",
        "Hybrid_NSN_SNN_rec": "SNN-Nonspiking (Network-to-Network)",
        "Hybrid_RNN_SNN_V1_same_layer": "SNN-ANN (Same Layer)",
        "Hybrid_NSN_SNN_V1_same_layer": "SNN-Nonspiking (Same Layer)",
        "old": "SNN (Evolutionary Team Implementation)",
    }

    try:
        con = sqlite3.connect("data/landscape-analysis.db")
        where_clause = " AND ".join([f"{feature} IS NOT NULL" for feature in features])
        variable_clause = " ,".join([f"a.{feature}" for feature in features])
        # SQL query to join two tables and filter for non-null/non-pending data
        query = f"""SELECT {variable_clause},
                          b.randman_id, b.model_type, b.recurrent, b.percent_snn
                   FROM randman_loss_surfaces a
                   JOIN randman_problems b on a.problem_id = b.id
                   WHERE {where_clause} {other_clause}"""
        print(query)
        df = pd.read_sql_query(query, con)
    except sqlite3.Error as e:
        print(f"Database error: {e}")
        return
    finally:
        if con:
            con.close()

    print(df)
    
    # Drop any remaining NaN values from the feature columns
    # df = df.dropna()

    df_clean = df.dropna(subset=features)
    scaled = StandardScaler().fit_transform(df_clean[features].values)
    pca = PCA(n_components=3)
    embedding = pca.fit_transform(scaled)
    # Adjust percent_snn to show 50% if it's hybrid and NaN
    df_clean['percent_snn'] = df_clean.apply(
        lambda row: 0.5 if pd.isna(row['percent_snn']) and 'Hybrid' in row['model_type'] else row['percent_snn'],
        axis=1
    )

    df_clean['combined_label'] = df_clean.apply(
        lambda row: f"ID: {row['randman_id']}\n"
                    f"{remapping.get(row['model_type'], row['model_type'])}\n"
                    f"Percent SNN: {row['percent_snn']}\n"
                    f"({'Recurrent' if row['recurrent'] == 1 else 'Non Recurrent'})", axis=1
    )

    unique_labels = df_clean['combined_label'].unique()
    num_labels = len(unique_labels)
    colors = plt.cm.get_cmap('tab20', num_labels)
    label_to_color = {label: colors(i) for i, label in enumerate(unique_labels)}
    color_array = [label_to_color[label] for label in df_clean['combined_label']]

    # 2D plot: PC1 vs PC2
    fig, ax = plt.subplots(figsize=(10, 8))
    ax.scatter(embedding[:, 0], embedding[:, 1], s=5, c=color_array)
    legend_elements = [
        plt.Line2D([0], [0], marker='o', color='w', label=label,
                   markerfacecolor=label_to_color[label], markersize=8)
        for label in unique_labels
    ]
    ax.legend(handles=legend_elements, title="Problem/Model Combination", bbox_to_anchor=(1.05, 1), loc='upper left', fontsize=8, ncol=1)
    plt.title('PCA Projection (PC1 vs PC2)', fontsize=14)
    plt.xlabel('PCA Component 1', fontsize=12)
    plt.ylabel('PCA Component 2', fontsize=12)
    plt.tight_layout(rect=[0, 0, 0.75, 1])
    plt.show()

    # 2D plot: PC3 vs PC4
    fig, ax = plt.subplots(figsize=(10, 8))
    ax.scatter(embedding[:, 1], embedding[:, 2], s=5, c=color_array)
    ax.legend(handles=legend_elements, title="Problem/Model Combination", bbox_to_anchor=(1.05, 1), loc='upper left', fontsize=8, ncol=1)
    plt.title('PCA Projection (PC2 vs PC3)', fontsize=14)
    plt.xlabel('PCA Component 2', fontsize=12)
    plt.ylabel('PCA Component 3', fontsize=12)
    plt.tight_layout(rect=[0, 0, 0.75, 1])
    plt.show()

    # 3D plot: PC1, PC2, PC4
    fig = plt.figure(figsize=(10, 8))
    ax = fig.add_subplot(111, projection='3d')
    ax.scatter(embedding[:, 0], embedding[:, 1], embedding[:, 2], s=10, c=color_array)
    ax.set_title('PCA Projection (3D: PC1, PC2, PC3)', fontsize=14)
    ax.set_xlabel('PCA Component 1')
    ax.set_ylabel('PCA Component 2')
    ax.set_zlabel('PCA Component 3')
    for i, label in enumerate(unique_labels):
        ax.scatter([], [], [], color=label_to_color[label], label=label)
    ax.legend(title="Problem/Model Combination", bbox_to_anchor=(1.05, 1), loc='upper left', fontsize=8, ncol=1)
    plt.tight_layout(rect=[0, 0, 0.75, 1])
    plt.show()

    # Variance explained
    explained_variance = pca.explained_variance_ratio_
    cumulative_variance = explained_variance.cumsum()
    plt.plot(range(1, len(cumulative_variance) + 1), cumulative_variance, marker='o', linestyle='--')
    plt.title("Cumulative Variance Explained by PCA Components")
    plt.xlabel("Number of PCA Components")
    plt.ylabel("Cumulative Variance Explained Ratio")
    plt.xticks(range(1, len(cumulative_variance) + 1))
    plt.grid()
    plt.show()

    # PCA projection matrix
    projection_matrix = pca.components_
    print("\nPCA Projection Matrix:")
    for i, component in enumerate(projection_matrix):
        print(f"Component {i + 1}: {component}")

plot_randman_pca(other_clause="and b.recurrent = 1 and b.dim=500 and b.model_type not like '%NSN%'")
```

```python
import os
import re
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
from PIL import Image

# Define the base path for your images
def get_png_files(BASE_PATH):
    # Get a sorted list of all PNG files
    png_files = []
    for root, dirs, files in os.walk(BASE_PATH):
        for f in files:
            if f.endswith('.png'):
                png_files.append(os.path.relpath(os.path.join(root, f), BASE_PATH))
    return sorted(png_files)
    
def plot_png_files(BASE_PATH):
    png_files = get_png_files(BASE_PATH)

    # Determine the grid size for the subplots
    n_files = len(png_files)
    n_cols = min(5, n_files)
    n_rows = (n_files + n_cols - 1) // n_cols

    # Create the figure and subplots
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(4 * n_cols, 4 * n_rows))
    axes = axes.flatten() if n_files > 1 else [axes]

    # Set a target size for the images to reduce memory usage
    # Adjust this value as needed to fit within your system's memory limits
    TARGET_SIZE = (256, 256) # 256x256 pixels

    # Loop through each image file and plot it
    for i, png_file in enumerate(png_files):
        full_path = os.path.join(BASE_PATH, png_file)
        
        # Open the image with PIL and resize it
        with Image.open(full_path) as pil_img:
            resized_img = pil_img.resize(TARGET_SIZE, Image.Resampling.LANCZOS)
            
        # Convert the resized image to a NumPy array for Matplotlib
        ax = axes[i]
        ax.imshow(resized_img)
        
        # Extract the seed value for the title
        match = re.search(r'seed_(\d+)', png_file)
        seed_label = f"seed_{match.group(1)}" if match else png_file
        ax.set_title(seed_label)
        ax.axis('off')

    # Hide any unused subplots
    for j in range(len(png_files), len(axes)):
        axes[j].axis('off')

    plt.tight_layout()
    plt.show()
```
