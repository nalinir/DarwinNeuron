### Visualization

```python
import sqlite3, numpy as np, pandas as pd
import matplotlib.pyplot as plt

def plot_feature():
    # Connect to the SQLite database
    con = sqlite3.connect("data/landscape-analysis.db")
    query = "SELECT * FROM randman_loss_surfaces WHERE ic_h_max IS NOT NULL AND ic_h_max != 'pending'"
    df = pd.read_sql_query(query, con)
    con.close()
    
    # Choose the feature to visualize
    feature = 'ic_eps_s'

    # Get unique problems
    problems = sorted(df['problem_id'].unique())

    # Prepare plot
    fig, axes = plt.subplots(len(problems), 1, figsize=(10, len(problems) * 2), sharex=False)
    
    # Handle case when there's only one problem
    if len(problems) == 1:
        axes = [axes]
    
    # Plot histograms per problem
    for i, problem in enumerate(problems):
        problem_data = df[df['problem_id'] == problem]
        
        # Calculate bins specific to this problem's data
        min_val = df[feature].min()
        max_val = df[feature].max()
        bins = np.linspace(min_val, max_val, 40)
        
        ax = axes[i]
        ax.hist(problem_data[feature], bins=bins, orientation='vertical')
        ax.set_yticks([])  # Hide y-axis ticks for compactness
        ax.set_ylabel(f"Problem {problem}", rotation=0, labelpad=30, va='center')
        
        # Add statistics to help compare
        mean_val = problem_data[feature].mean()
        ax.axvline(mean_val, color='red', linestyle='dashed', linewidth=1, label=f'Mean: {mean_val:.3f}')
        
        ax.spines['top'].set_visible(False)
        ax.spines['right'].set_visible(False)
        ax.spines['left'].set_visible(False)
        ax.legend()

    # Aesthetics
    axes[-1].set_xlabel(feature)
    plt.suptitle(f"Distribution of {feature} by Problem", fontsize=14)
    plt.tight_layout(rect=[0, 0, 1, 0.96])
    plt.show()
plot_feature()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import sqlite3
import numpy as np

def plot_multiple_features():
    """
    Connects to a SQLite database, queries data, and plots feature histograms
    for different problems in a grid format.

    The plots are organized with each row representing a unique problem, and
    each column representing a feature. The y-axis labels are dynamically
    set to indicate the Randman ID, model type, and whether it's 'Recurrent'
    or 'Non Recurrent'.
    """
    # Connect to the SQLite database
    try:
        con = sqlite3.connect("data/landscape-analysis.db")
        # SQL query to join two tables and filter for non-null/non-pending data
        query = """SELECT a.*, b.randman_id, b.model_type, b.recurrent FROM randman_loss_surfaces a
                   JOIN randman_problems b on a.problem_id = b.id 
                   WHERE ic_h_max IS NOT NULL AND ic_h_max != 'pending'"""
        df = pd.read_sql_query(query, con)
    except sqlite3.Error as e:
        print(f"Database error: {e}")
        return
    finally:
        if con:
            con.close()

    # Define features to visualize
    features = ['ic_h_max', 'ic_eps_s', 'ic_eps_max', 'ic_eps_ratio', 'ic_m0']

    # Get unique combinations of randman_id, model_type, and recurrent status
    problems = df[['randman_id', 'model_type', 'recurrent']].drop_duplicates().sort_values(by=['randman_id', 'model_type', 'recurrent']).values.tolist()

    # Define a remapping dictionary for model type labels
    remapping = {
        "SNN": "SNN",
        "ANN_with_LIF_output": "ANN",
        "NSN_with_LIF_output": "Nonspiking",
        "Hybrid_RNN_SNN_rec": "SNN-ANN (Network-to-Network)",
        "Hybrid_NSN_SNN_rec": "SNN-Nonspiking (Network-to-Network)",
        "Hybrid_RNN_SNN_V1_same_layer": "SNN-ANN (Same Layer)",
        "Hybrid_NSN_SNN_V1_same_layer": "SNN-Nonspiking (Same Layer)",
    }
    
    # Create subplot grid with shared x-axes for columns
    fig, axes = plt.subplots(len(problems), len(features), 
                             figsize=(len(features) * 3, len(problems) * 1.5), 
                             sharex='col')
    
    # Handle the case where there is only one problem to ensure `axes` is a 2D array
    if len(problems) == 1:
        axes = [axes]
    
    # Plot histograms for each problem and feature combination
    for i, (randman_id, model_type, recurrent) in enumerate(problems):
        # Correctly filter the DataFrame for the current unique combination of randman_id, model_type, and recurrent
        problem_data = df[(df['randman_id'] == randman_id) & 
                          (df['model_type'] == model_type) & 
                          (df['recurrent'] == recurrent)]
        
        # Find the remapped model type name based on a prefix match
        remapped_model_type = model_type
        for prefix, new_label in remapping.items():
            if model_type.startswith(prefix):
                remapped_model_type = new_label
                break
        
        # Create the conditional label for the y-axis
        recurrent_label = "Recurrent" if recurrent == 1 else "Non Recurrent"
        
        # Updated row_label to be three lines for better spacing
        row_label = (f"Randman ID: {randman_id}\n"
                     f"{remapped_model_type}\n"
                     f"({recurrent_label}, n={len(problem_data)})")

        for j, feature in enumerate(features):
            ax = axes[i][j]
            
            # Prepare data, dropping NaNs
            problem_feature_data = problem_data[feature].dropna()
            
            # Check if all values for this feature in this problem are the same
            if len(problem_feature_data.unique()) <= 1 and not problem_feature_data.empty:
                unique_val = problem_feature_data.iloc[0]
                # Plot a vertical line and add text for single-value data
                ax.axvline(x=unique_val, color='r', linestyle='--')
                ax.text(unique_val, ax.get_ylim()[1] * 0.8,
                        f'{unique_val:.2f}',
                        ha='center', va='top', rotation=90, fontsize=8, color='r')
            else:
                # Use np.linspace to create consistent bins across all plots for a given feature
                bins = np.linspace(df[feature].min(), df[feature].max(), 40)
                ax.hist(problem_feature_data, bins=bins, orientation='vertical')
            
            # Hide y-axis ticks and spines for a cleaner look
            ax.set_yticks([])
            ax.spines['top'].set_visible(False)
            ax.spines['right'].set_visible(False)
            ax.spines['left'].set_visible(False)
            
            # Set labels only for the first column and last row
            if j == 0:
                ax.set_ylabel(row_label, rotation=0, labelpad=90, va='center')
            
            if i == len(problems) - 1:
                ax.set_xlabel(feature, fontsize=9)
    
    # Add a main title and adjust the layout to prevent overlap
    plt.suptitle("Feature Distributions by Randman Dataset and Model Type", fontsize=14)
    # Adjust subplot params to make space for the longer y-axis labels
    plt.subplots_adjust(left=0.4)
    plt.tight_layout(rect=[0, 0, 1, 0.96])
    plt.show()

# Run the function to generate the plot
plot_multiple_features()
```

```python
import sqlite3, umap
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
sns.set_theme()
def plot_umap():
    con = sqlite3.connect("data/landscape-analysis.db")
    df = pd.read_sql_query("SELECT problem_id, ic_h_max, ic_eps_s, ic_eps_max, ic_eps_ratio, ic_m0 FROM randman_loss_surfaces", con)
    con.close()
    
    df= df.dropna()
    scaled = StandardScaler().fit_transform(df[['ic_h_max', 'ic_eps_s', 'ic_eps_max', 'ic_eps_ratio', 'ic_m0']].values)
    reducer = umap.UMAP()
    embedding = reducer.fit_transform(scaled)
    scatter = plt.scatter(embedding[:, 0], embedding[:, 1], s=5, 
                          c=df['problem_id'], cmap='tab10')
    legend1 = plt.legend(*scatter.legend_elements(), title="Problem ID", fontsize=6)
    plt.gca().add_artist(legend1)

    plt.title('UMAP Projection', fontsize=7)
    plt.xlabel('UMAP Dimension 1')
    plt.ylabel('UMAP Dimension 2')
    plt.show()
    
plot_umap()
```

```python
import sqlite3
import umap
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# Set a consistent theme for the plot
sns.set_theme()

def plot_umap():
    """
    Connects to a SQLite database, queries data, and performs UMAP
    dimensionality reduction. The resulting 2D plot is colored by
    a unique combination of Randman ID, model type, and recurrent status.
    """
    # Connect to the SQLite database
    try:
        con = sqlite3.connect("data/landscape-analysis.db")
        # SQL query to join two tables and filter for non-null/non-pending data
        query = """SELECT a.ic_h_max, a.ic_eps_s, a.ic_eps_max, a.ic_eps_ratio, a.ic_m0, 
                          b.randman_id, b.model_type, b.recurrent
                   FROM randman_loss_surfaces a
                   JOIN randman_problems b on a.problem_id = b.id 
                   WHERE ic_h_max IS NOT NULL AND ic_h_max != 'pending'"""
        df = pd.read_sql_query(query, con)
    except sqlite3.Error as e:
        print(f"Database error: {e}")
        return
    finally:
        if con:
            con.close()
    
    # Drop any remaining NaN values from the feature columns
    df = df.dropna()

    # Define the features for UMAP
    features = ['ic_h_max', 'ic_eps_s', 'ic_eps_max', 'ic_eps_ratio', 'ic_m0']

    # Standardize the features before applying UMAP
    scaled_data = StandardScaler().fit_transform(df[features].values)

    # Apply UMAP dimensionality reduction
    reducer = umap.UMAP()
    embedding = reducer.fit_transform(scaled_data)

    # Define a remapping dictionary for model type labels
    remapping = {
        "SNN": "SNN",
        "ANN_with_LIF_output": "ANN",
        "NSN_with_LIF_output": "Nonspiking",
        "Hybrid_RNN_SNN_rec": "SNN-ANN (Network-to-Network)",
        "Hybrid_NSN_SNN_rec": "SNN-Nonspiking (Network-to-Network)",
        "Hybrid_RNN_SNN_V1_same_layer": "SNN-ANN (Same Layer)",
        "Hybrid_NSN_SNN_V1_same_layer": "SNN-Nonspiking (Same Layer)",
    }

    # Create a new column to store the combined label for each data point
    df['combined_label'] = df.apply(lambda row: f"ID: {row['randman_id']}\n"
                                                f"{remapping[row['model_type']]}\n"
                                                f"({ 'Recurrent' if row['recurrent'] == 1 else 'Non Recurrent'})", axis=1)

    # Get a list of unique labels to use for coloring
    unique_labels = df['combined_label'].unique()
    num_labels = len(unique_labels)

    # Create a color map and assign a color to each unique label
    colors = plt.cm.get_cmap('tab20', num_labels)
    label_to_color = {label: colors(i) for i, label in enumerate(unique_labels)}

    # Create a color array for the scatter plot
    color_array = [label_to_color[label] for label in df['combined_label']]

    # Create the plot
    fig, ax = plt.subplots(figsize=(10, 8))
    scatter = ax.scatter(embedding[:, 0], embedding[:, 1], s=5, c=color_array)
    
    # Create a custom legend with the combined labels
    legend_elements = [
        plt.Line2D([0], [0], marker='o', color='w', label=label,
                   markerfacecolor=label_to_color[label], markersize=8)
        for label in unique_labels
    ]
    ax.legend(handles=legend_elements, title="Problem/Model Combination", loc='best', fontsize=8, ncol=2)

    # Add titles and labels
    plt.title('UMAP Projection', fontsize=14)
    plt.xlabel('UMAP Dimension 1', fontsize=12)
    plt.ylabel('UMAP Dimension 2', fontsize=12)
    plt.show()
    
plot_umap()
```
